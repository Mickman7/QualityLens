```python
import cv2
import os
import shutil
```

```python
def is_blurry(image_path, threshold=100.0):
    image = cv2.imread(image_path)
    if image is None: 
        return True
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    variance = cv2.Laplacian(gray, cv2.CV_64F).var()
    return variance < threshold

# Define paths
RAW_DIR = "../data/raw"
PROCESSED_DIR = "../data/processed"
categories = ['fresh', 'rotten']

# Global counters for the final summary
total_sharp = 0
total_blurry = 0

for cat in categories:
    target_folder = os.path.join(PROCESSED_DIR, cat)
    source_folder = os.path.join(RAW_DIR, cat)
    os.makedirs(target_folder, exist_ok=True)
    
    # Category-specific counters
    cat_sharp = 0
    cat_blurry = 0
    
    print(f"\n--- Processing: {cat.upper()} ---")
    
    for filename in os.listdir(source_folder):
        if filename.startswith('.'): continue
        path = os.path.join(source_folder, filename)
        
        if not is_blurry(path):
            shutil.copy(path, os.path.join(target_folder, filename))
            cat_sharp += 1
        else:
            cat_blurry += 1

    print(f"{cat.capitalize()} category done.")
    print(f"   - Sharp images kept: {cat_sharp}")
    print(f"   - Blurry images skipped: {cat_blurry}")
    
    total_sharp += cat_sharp
    total_blurry += cat_blurry

# Final Project Summary
print("\n" + "="*30)
print("       CLEANING SUMMARY")
print("="*30)
print(f"Total Sharp Images:  {total_sharp}")
print(f"Total Blurry Images: {total_blurry}")
print(f"Final Dataset Size:  {total_sharp}")
print(f"Discard Rate:        {(total_blurry / (total_sharp + total_blurry) * 100):.1f}%")
print("="*30)
```
